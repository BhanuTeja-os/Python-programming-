```python
def reverse(a):
  return int(str(a)[::-1])
a=input()
result = reverse(a)
print(result)
```

```python
#program that reverse a string using functions
def rev(s):
  x=""
  for i in range(len(s)-1,-1,-1):
      x=x+s[i]
  return x
print(rev("hi"))
```

```python
def palindrome(s):
   return 1 if s == s[::-1] else 0
s=input()
print(palindrome(s))
```

```python
def multiplefunction(*num):
   s = sum(num)
   c = len(num)
   avg = s/c
   return s,c,avg
res_sum,res_count,res_avg = multiplefunction(1,2,3,4,5,6,7,8)
print(f"the sum of numbers is {res_sum},the count is {res_count},the average is {res_avg :.2f}")
```
