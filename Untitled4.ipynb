```python
#operations on set
s = {1,2,5,8,6}
s1 = {2,4,7,9}
print(s|s1)
print(s&s1)
print(s-s1)
print(s^s1)
```

```python
s.add(4)
s.update((7,9,10,32))
s.remove(32)
s.discard("new")
s.pop()
s.union(s1)
print(s)
s.intersection(s1)
print(s)
s1.difference(s)
print(s1)
s.symmetric_difference(s1)
print(s)
```
